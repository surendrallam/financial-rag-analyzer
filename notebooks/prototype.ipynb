## Testing the Parser (LlamaParse)

```python
from llama_parse import LlamaParse

# Initialize parser
parser = LlamaParse(
    api_key=os.getenv("LLAMA_CLOUD_API_KEY"),
    result_type="markdown"
)

# Test on a single file (Ensure you have a dummy pdf in ../data)
documents = parser.load_data("../data/my_statement.pdf")

# Inspect the output - crucial to see if tables are preserved!
print(documents[0].text[:1000])
```

## Testing Embedding & Querying

```python
from llama_index.core import VectorStoreIndex, Settings
from llama_index.llms.openai import OpenAI

# Setup
Settings.llm = OpenAI(model="gpt-4o")

# Build Index
index = VectorStoreIndex.from_documents(documents)
query_engine = index.as_query_engine()

# Test a complex query
response = query_engine.query("What is the total sum of all 'Debit' transactions?")
print(str(response))
```
